## Tipos de Avaliadores

### Avaliadores integrados

O AgentCore fornece 13 avaliadores pré-configurados que usam Modelos de Linguagem de Grande Escala (LLMs) como juízes para avaliar o desempenho do agente. Esses avaliadores vêm com templates de prompt cuidadosamente elaborados e modelos avaliadores pré-selecionados, padronizando os critérios de avaliação em diferentes casos de uso. Eles estão prontos para uso e você não precisa adicionar nenhuma configuração adicional para começar.

Esses avaliadores são divididos em 4 grupos diferentes:

- **Qualidade da resposta**: avaliadores que ajudam você a decidir se seu agente está funcionando como esperado em cada turno. Eles funcionam por trace e avaliam cada interação usuário-agente.
- **Conclusão de tarefas**: esta categoria tem um avaliador (Sucesso do objetivo) que avaliará a sessão como um todo. Em uma conversa multi-turno, este avaliador ajuda você a decidir se o objetivo (ou objetivos) do usuário foram completados e se o resultado final foi alcançado. Em situações onde o agente faz perguntas de acompanhamento, este avaliador é essencial para entender se as tarefas solicitadas foram realmente completadas.
- **Nível de ferramenta**: avaliadores que ajudam você a entender quão bem-sucedida é a chamada de ferramentas. Ele mede a precisão da seleção de ferramentas e parâmetros do agente no nível da ferramenta. Se um agente está chamando duas ou mais ferramentas diferentes em um único turno, cada uma delas terá sua própria métrica no trace.
- **Segurança**: avaliadores que detectam se seu agente está sendo prejudicial ou fazendo generalizações estereotipadas sobre indivíduos ou grupos.

Ao usar métricas integradas, tudo é tratado para você, do prompt ao modelo. Isso significa que seu avaliador não pode ser modificado, para manter a avaliação consistente e confiável entre todos os usuários. Você pode, no entanto, criar suas próprias métricas usando uma métrica integrada como base. Para isso, fornecemos os **Templates de Prompt** para nossos avaliadores integrados.

### Avaliadores personalizados

Os avaliadores personalizados fornecem máxima flexibilidade, permitindo que você defina cada aspecto do seu processo de avaliação enquanto utiliza LLMs como juízes subjacentes. Você pode personalizar o seguinte no seu avaliador personalizado:

- **Modelo avaliador**: Escolha o LLM que melhor se adapta às suas necessidades de avaliação
- **Prompts de avaliação**: Elabore instruções de avaliação específicas para seu caso de uso
- **Esquema de pontuação**: Projete sistemas de pontuação alinhados com as métricas da sua organização

### Gerando traces no AgentCore Observability a partir de um agente

O AgentCore Observability fornece visibilidade abrangente do comportamento do agente durante invocações, aproveitando traces [OpenTelemetry (OTEL)](https://opentelemetry.io/) como base para capturar e estruturar dados detalhados de execução. O AgentCore depende do [AWS Distro for OpenTelemetry (ADOT)](https://aws-otel.github.io/) para instrumentar diferentes tipos de traces OTEL em vários frameworks de agentes.

Quando seu agente está hospedado no AgentCore Runtime (como nosso agente neste tutorial), a instrumentação do AgentCore Observability é automática, com configuração mínima. Tudo que você precisa fazer é incluir `aws-opentelemetry-distro` no `requirements.txt` e o AgentCore Runtime lida com a configuração OTEL automaticamente. Quando seu agente não está rodando no AgentCore Runtime, você precisará instrumentá-lo com ADOT para tê-lo disponível no AgentCore Observability. Você precisa configurar variáveis de ambiente para direcionar dados de telemetria ao CloudWatch e executar seu agente com instrumentação OpenTelemetry.

O processo é o seguinte:

![session_traces](../images/observability_traces.png)

Uma vez que os traces da sua sessão estejam disponíveis no AgentCore Observability, você pode usar o AgentCore Evaluations para avaliar o comportamento do seu agente.

### Níveis de avaliação
O AgentCore Evaluations opera em diferentes níveis da interação do agente. Você pode analisar a conversa de ida e volta como um todo usando as informações da sessão. Você também pode avaliar a resposta do agente a uma pergunta do usuário em um turno individual da conversa usando as informações do trace. Ou você pode avaliar informações dentro de um turno, que incluem a chamada de ferramentas e seleção de parâmetros, usando os dados do span.

Você pode criar métricas personalizadas para os diferentes níveis. As métricas integradas operam no seguinte escopo:

![nível de métricas](../images/metrics_per_level.png)

Neste tutorial, criaremos uma métrica no nível de trace.

### Criar avaliador personalizado

Vamos agora criar uma métrica personalizada para qualidade de resposta que nos permitirá ter a escala de 5 níveis. Para isso, precisamos selecionar um modelo avaliador, fornecer instruções para a avaliação e definir a escala de classificação. Neste caso, nossa escala irá de Muito Bom a Muito Ruim. Vamos recuperar a configuração de avaliação do arquivo JSON

```python
with open("metric.json") as f:
    print("Reading custom metric details")
    eval_config = json.load(f)
eval_config
```

Podemos então usar o método `create_evaluator` para criar o avaliador. Criaremos o avaliador `response_Quality` no nível `TRACE`.

Ao criar seu avaliador personalizado, você pode escolher aplicá-lo no nível de chamada de ferramenta, nível de trace ou nível de sessão.

* **Uma chamada de ferramenta** é um span que representa a invocação de uma função externa, API ou capacidade pelo agente. Spans de chamada de ferramenta tipicamente capturam informações como o nome da ferramenta, parâmetros de entrada, tempo de execução e saída. Os detalhes da chamada de ferramenta são usados para avaliar se o agente selecionou e usou as ferramentas corretamente e eficientemente.

* **Um trace** é um registro completo de uma única execução ou solicitação do agente. Um trace contém um ou mais spans, que representam as operações individuais realizadas durante essa execução. Traces fornecem visibilidade de ponta a ponta nas decisões do agente e uso de ferramentas.

* **Uma sessão** representa um agrupamento lógico de interações relacionadas de um único usuário ou fluxo de trabalho. Uma sessão pode conter um ou mais traces. Sessões ajudam você a visualizar e avaliar o comportamento do agente em interações de múltiplos passos, em vez de focar em solicitações individuais.

```python
custom_evaluator = eval_client.create_evaluator(
    name="response_quality_for_scope",
    level="TRACE",
    description="Response quality evaluator",
    config=eval_config
)
```
