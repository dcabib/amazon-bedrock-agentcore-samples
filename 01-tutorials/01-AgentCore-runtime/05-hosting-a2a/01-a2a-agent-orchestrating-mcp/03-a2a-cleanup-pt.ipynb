```python
print("🛡️  Iniciando limpeza de segurança...")
import json
try:
    # bedrock_client = boto3.client("bedrock", region_name=REGION)
    
    # Deletar função de execução
    print("  🗑️  Deletando função de execução do Agente 1 - AWS Docs...")
    delete_agentcore_runtime_execution_role(AWS_DOCS_ROLE_NAME)
    print("  ✅ Função de execução deletada")

    print("  🗑️  Deletando função de execução do Agente 2 - AWS Blogs...")
    delete_agentcore_runtime_execution_role(AWS_BLOG_ROLE_NAME)
    print("  ✅ Função de execução deletada")

    print("  🗑️  Deletando função de execução do Orquestrador...")
    delete_agentcore_runtime_execution_role(ORCHESTRATOR_ROLE_NAME)
    print("  ✅ Função de execução deletada")
    
    # Limpar Cognito e secrets
    print("  🗑️  Limpando recursos do Cognito...")
    cs = json.loads(get_cognito_secret())
    cleanup_cognito_resources(cs['pool_id'])
    print("  ✅ Recursos do Cognito limpos")
    
    print("  🗑️  Deletando secret de suporte ao cliente...")
    delete_cognito_secret()
    print("  ✅ Secret de suporte ao cliente deletado")

    print("  🗑️  Deletando parâmetro SSM...")
    delete_ssm_parameter(SSM_DOCS_AGENT_ARN)
    print("  ✅ Parâmetro SSM deletado")

    print("  🗑️  Deletando parâmetro SSM...")
    delete_ssm_parameter(SSM_BLOGS_AGENT_ARN)
    print("  ✅ Parâmetro SSM deletado")
    
except Exception as e:
    print(f"  ⚠️  Erro durante limpeza de segurança: {e}")
```
