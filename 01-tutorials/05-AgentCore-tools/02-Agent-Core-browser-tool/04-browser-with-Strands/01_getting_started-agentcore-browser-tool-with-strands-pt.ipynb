## Etapa 2: Importar Bibliotecas

````python
from strands import Agent
from strands_tools.browser import AgentCoreBrowser
import time

print("Bibliotecas importadas com sucesso!")
````

## Etapa 3: Criar Agente com Configurações Otimizadas

````python
def create_agent():
    """Criar e configurar o agente Strands com AgentCoreBrowser"""
    # Inicializar a ferramenta oficial AgentCoreBrowser
    agent_core_browser = AgentCoreBrowser(region="us-west-2")
    
    # Criar agente com modelo Claude Haiku e configurações otimizadas
    agent = Agent(
        tools=[agent_core_browser.browser],
        model="global.anthropic.claude-haiku-4-5-20251001-v1:0",
        system_prompt="""Você é um analista financeiro inteligente especializado em analisar sites de ações e financeiros. Quando solicitado a analisar um site financeiro:

1. Use a ferramenta de navegador para visitar e interagir com o site de forma EFICIENTE
2. Concentre-se em extrair informações financeiras chave RAPIDAMENTE:

**Para Sites Financeiros/de Ações (MarketWatch, Bloomberg, etc.):**
- Preços atuais de ações e dados de mercado
- Movimentos de preços e tendências (mudanças diárias, semanais, mensais)
- Métricas e índices financeiros chave (P/E, Valor de Mercado, etc.)
- Volume de negociação e atividade de mercado
- Notícias recentes e sentimento do mercado
- Recomendações de analistas e metas de preço
- Fundamentos da empresa e indicadores de desempenho

IMPORTANTE: Trabalhe eficientemente e complete sua análise dentro de 2-3 interações do navegador. Sempre forneça insights financeiros específicos e acionáveis com números e pontos de dados reais. Seja conciso, mas abrangente em sua análise, focando nas métricas financeiras mais importantes para investidores."""
    )
    return agent

# Inicializar agente globalmente
strands_agent = create_agent()

print("Agente Strands criado com integração AgentCoreBrowser!")
````

## Etapa 4: Criar Função Invoke

````python
def invoke(payload):
    """Função invoke estruturada para interação com o agente"""
    user_message = payload.get("prompt", "")
    
    try:
        print("Iniciando análise...")
        start_time = time.time()
        
        response = strands_agent(user_message)
        
        elapsed_time = time.time() - start_time
        print(f"Análise concluída em {elapsed_time:.2f} segundos")
        
        return response.message["content"][0]["text"] if response.message.get("content") else str(response)
        
    except Exception as e:
        print(f"Erro durante a análise: {str(e)}")
        return f"Ocorreu um erro: {str(e)}"

print("Função invoke criada!")

# Nota: Se você acessar sites lentos que causem timeouts, pode ser necessário
# implementar tratamento de timeout em seu código para evitar operações travadas.
````

## Etapa 5: Testar Análise Financeira do MarketWatch

````python
# Testar com a página da Tesla no MarketWatch
financial_test = {
    "prompt": "Por favor, analise a página de ações da Tesla em https://www.marketwatch.com/investing/stock/tsla e forneça insights financeiros chave"
}

print("Testando Análise Financeira do MarketWatch")
print("=" * 50)

result = invoke(financial_test)
print(result)

print("\nTeste de análise do MarketWatch concluído!")
````

## Etapa 6: Função de Análise Interativa

````python
def analyze_website(url, question=None):
    """Função conveniente para análise interativa de sites"""
    if question:
        prompt = f"Visite {url} e responda esta pergunta: {question}"
    else:
        prompt = f"Por favor, analise o site em {url} e forneça insights financeiros abrangentes"
    
    payload = {"prompt": prompt}
    return invoke(payload)

# Exemplo de uso
print("Função de Análise Interativa Pronta!")
print("\nExemplo de uso:")
print('analyze_website("https://www.marketwatch.com/investing/stock/aapl", "Qual é o desempenho atual das ações da Apple?")')
print("\nFunção de análise interativa pronta!")
````

## O Que Aconteceu nos Bastidores

Quando você executa este notebook, o Agente Strands é inicializado com o modelo Claude Haiku e se conecta ao serviço de automação de navegador AgentCore da AWS Bedrock. Ao receber sua solicitação de análise financeira, o agente navega inteligentemente para MarketWatch.com, extrai dados de ações em tempo real, incluindo preços, capitalização de mercado, índices P/E e volumes de negociação, e então processa essas informações através da IA do Claude para gerar insights financeiros abrangentes.

O rastreamento de desempenho mede a velocidade de execução (tipicamente 14-16 segundos), e o tratamento de erros gerencia graciosamente quaisquer problemas de rede ou bloqueios de sites. Todo o processo é orquestrado através do padrão da função invoke(), que fornece uma interface limpa enquanto gerencia automação complexa do navegador e processamento de IA nos bastidores.

O resultado é o que parece ser uma simples chamada de função que na verdade coordena múltiplos serviços AWS, modelos de IA e web scraping para fornecer análise financeira confiável.
